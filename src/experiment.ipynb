## This is a notebook for experiments

```python
import numpy as np 
import pandas as pd

dataset=pd.read_csv("../data/loan_approval_dataset.csv")
dataset.head()
```

```python
dataset.columns
```

```python
dataset[" education"].unique()
```

```python
dataset.drop(columns=["loan_id"],inplace=True)
```

```python
dataset.describe()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder=OrdinalEncoder(categories=[[" Graduate"," Not Graduate"]])

dataset[" education"]=encoder.fit_transform(dataset[[" education"]])
```

```python
dataset[" self_employed"] = dataset[" self_employed"].map({
    " Yes": 0,
    " No": 1
}).astype(int)

dataset[" loan_status"] = dataset[" loan_status"].map({
    " Approved": 0,
    " Rejected": 1
}).astype(int)
```

```python
dataset
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

input_data=dataset.iloc[:,:-1]
ouptut_data=dataset.iloc[:,-1]
xtrain,xeval,train,yeval=train_test_split(input_data,ouptut_data,stratify=ouptut_data,test_size=0.30,random_state=15)
xval,xtest,yval,ytest=train_test_split(xeval,yeval,test_size=0.50,stratify=yeval,random_state=15)

scaler=StandardScaler()
xtrain_scaled=scaler.fit_transform(xtrain)
xval_scaled=scaler.fit(xval)
xtest_scaled=scaler.fit(xtest)
```
